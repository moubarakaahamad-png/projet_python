## 1. Calcul numérique et flottants

L'ordinateur arrondit les nombres décimaux. Il ne faut donc pas comparer deux décimaux avec `==`.

```python
print(0.1 + 0.2 == 0.3)

def presque_egaux(a, b):
    return abs(a - b) < 0.000001

print(presque_egaux(0.1 + 0.2, 0.3))

total = 0
for i in range(1000):
    total = total + 0.1
print(total)
print(abs(total - 100))
```

## 2. Arithmétique et nombres premiers

Le PGCD, un test pour savoir si un nombre est premier, et la liste des nombres premiers jusqu'à 50.

```python
def pgcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def est_premier(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True

print(pgcd(1071, 462))
print(est_premier(97))
print(est_premier(100))

premiers = []
for n in range(50):
    if est_premier(n):
        premiers.append(n)
print(premiers)
```

## 3. Polynômes et recherche de racines

Un polynôme est une liste de coefficients : `[-2, 0, 1]` veut dire $-2 + 0x + 1x^{2}$, donc $x^{2} - 2$. On cherche sa racine avec la méthode de Newton.

```python
def evaluer(P, x):
    resultat = 0
    for coef in reversed(P):
        resultat = resultat * x + coef
    return resultat

def deriver(P):
    resultat = []
    for i in range(1, len(P)):
        resultat.append(i * P[i])
    return resultat

def newton(P, depart):
    x = depart
    for i in range(20):
        x = x - evaluer(P, x) / evaluer(deriver(P), x)
    return x

P = [-2, 0, 1]
print(evaluer(P, 3))
print(deriver(P))
print(newton(P, 1))
```

## 4. Matrices et systèmes linéaires

On résout le système $2x + y = 5$ et $x + 3y = 10$. Chaque ligne de la matrice `A` est une équation, et `b` contient les résultats.

```python
import numpy as np

def produit(A, B):
    resultat = []
    for i in range(len(A)):
        ligne = []
        for j in range(len(B[0])):
            total = 0
            for k in range(len(B)):
                total = total + A[i][k] * B[k][j]
            ligne.append(total)
        resultat.append(ligne)
    return resultat

A = [[2, 1], [1, 3]]
b = [5, 10]

resultat = np.linalg.solve(np.array(A), np.array(b))
solution = [float(resultat[0]), float(resultat[1])]
print(solution)
print(produit(A, [[solution[0]], [solution[1]]]))
```

## 5. Géométrie du plan

La distance entre deux points, et l'aire d'un triangle grâce au déterminant.

```python
import math

def distance(A, B):
    return math.sqrt((B[0] - A[0]) ** 2 + (B[1] - A[1]) ** 2)

def aire_triangle(A, B, C):
    det = (B[0] - A[0]) * (C[1] - A[1]) - (B[1] - A[1]) * (C[0] - A[0])
    return abs(det) / 2

print(distance((0, 0), (3, 4)))
print(aire_triangle((0, 0), (4, 0), (0, 3)))
```
