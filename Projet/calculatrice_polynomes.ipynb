# Calculatrice de polynômes

Un polynôme est représenté par la liste de ses coefficients, du terme constant vers le plus haut degré. Par exemple `[2, 3, 0, 5]` représente $2 + 3x + 5x^{3}$.

Ce petit projet reprend des notions des TP : les listes et les boucles (TP1), la méthode de Horner et la dérivée (TP3), la recherche de racine par Newton (TP3).

Exécutez les cellules dans l'ordre avec Maj+Entrée.

## 1. Les fonctions

```python
def evaluer(P, x):
    resultat = 0
    for coef in reversed(P):
        resultat = resultat * x + coef
    return resultat

def deriver(P):
    resultat = []
    for i in range(1, len(P)):
        resultat.append(i * P[i])
    return resultat

def somme(P, Q):
    n = max(len(P), len(Q))
    resultat = []
    for i in range(n):
        a = 0
        b = 0
        if i < len(P):
            a = P[i]
        if i < len(Q):
            b = Q[i]
        resultat.append(a + b)
    return resultat

def produit_poly(P, Q):
    resultat = [0] * (len(P) + len(Q) - 1)
    for i in range(len(P)):
        for j in range(len(Q)):
            resultat[i + j] = resultat[i + j] + P[i] * Q[j]
    return resultat

def newton(P, depart, tolerance=1e-10):
    dP = deriver(P)
    x = depart
    for tours in range(1, 101):
        pente = evaluer(dP, x)
        if pente == 0:
            return None, tours
        nouveau = x - evaluer(P, x) / pente
        if abs(nouveau - x) < tolerance:
            return nouveau, tours
        x = nouveau
    return None, 100
```

## 2. L'affichage et le tracé

```python
import matplotlib.pyplot as plt

def afficher(P):
    morceaux = []
    for i in range(len(P)):
        coef = P[i]
        if coef == 0:
            continue
        if i == 0:
            morceaux.append(str(coef))
        else:
            if coef == 1:
                texte = "x"
            elif coef == -1:
                texte = "-x"
            else:
                texte = str(coef) + "x"
            if i > 1:
                texte = texte + "^" + str(i)
            morceaux.append(texte)
    if len(morceaux) == 0:
        return "0"
    return " + ".join(morceaux).replace("+ -", "- ")

def tracer(P, a, b):
    xs = []
    ys = []
    for i in range(201):
        x = a + (b - a) * i / 200
        xs.append(x)
        ys.append(evaluer(P, x))
    plt.plot(xs, ys)
    plt.axhline(0, color="black")
    plt.title("P(x) = " + afficher(P))
    plt.show()
```

## 3. Démonstration

Changez `P` et `Q` pour essayer vos propres polynômes.

```python
P = [2, 3, 0, 5]
Q = [1, 1]

print("P(x) =", afficher(P))
print("Q(x) =", afficher(Q))
print("P(2) =", evaluer(P, 2))
print("P'(x) =", afficher(deriver(P)))
print("P + Q =", afficher(somme(P, Q)))
print("P x Q =", afficher(produit_poly(P, Q)))
```

## 4. Chercher une racine

On cherche une solution de $x^{3} - 2x - 5 = 0$ avec la méthode de Newton, en partant de 2.

```python
R = [-5, -2, 0, 1]
racine, tours = newton(R, 2)
print("R(x) =", afficher(R))
print("racine :", racine)
print("tours :", tours)
print("R(racine) =", evaluer(R, racine))
```

## 5. Le graphique

La courbe coupe l'axe horizontal à la racine trouvée.

```python
tracer(R, -3, 3)
```
